## Connect Python with MySQL Database and Perform CRUD Operations

### Install MySQL Connector

```python
!pip install mysql-connector-python
```

### 1. Import MySQL Connector

```python
import mysql.connector
```

### 2. Connect Python with MySQL

```python
con = mysql.connector.connect(
    host="localhost",
    user="root",
    password="root"
)

print("MySQL Database Connected Successfully")
```

### 3. Create Database

```python
cursor = con.cursor()

cursor.execute("CREATE DATABASE IF NOT EXISTS python_crud")
print("Database Created Successfully")
```

### 4. Select Database and Create Table

```python
con.database = "python_crud"

cursor.execute("""
CREATE TABLE IF NOT EXISTS student (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(100),
    email VARCHAR(100),
    course VARCHAR(100)
)
""")

print("Student Table Created Successfully")
```

### 5. CREATE - Insert Records

```python
sql = "INSERT INTO student (name, email, course) VALUES (%s, %s, %s)"

data = [
    ("Darshil", "darshil@gmail.com", "Computer Engineering"),
    ("Rahul", "rahul@gmail.com", "Information Technology"),
    ("Amit", "amit@gmail.com", "Computer Engineering")
]

cursor.executemany(sql, data)
con.commit()

print(cursor.rowcount, "records inserted successfully")
```

### 6. READ - Display All Records

```python
cursor.execute("SELECT * FROM student")

records = cursor.fetchall()

for row in records:
    print(row)
```

### 7. READ - Display One Record

```python
cursor.execute("SELECT * FROM student WHERE id = %s", (1,))

record = cursor.fetchone()

print(record)
```

### 8. UPDATE - Update a Record

```python
sql = "UPDATE student SET name = %s, email = %s, course = %s WHERE id = %s"

data = ("Darshil Patel", "darshilpatel@gmail.com", "Computer Engineering", 1)

cursor.execute(sql, data)
con.commit()

print(cursor.rowcount, "record updated successfully")
```

### 9. READ - Check Updated Record

```python
cursor.execute("SELECT * FROM student WHERE id = %s", (1,))

record = cursor.fetchone()

print(record)
```

### 10. DELETE - Delete a Record

```python
sql = "DELETE FROM student WHERE id = %s"

data = (3,)

cursor.execute(sql, data)
con.commit()

print(cursor.rowcount, "record deleted successfully")
```

### 11. READ - Display Final Records

```python
cursor.execute("SELECT * FROM student")

records = cursor.fetchall()

for row in records:
    print(row)
```

### 12. Close Database Connection

```python
cursor.close()
con.close()

print("MySQL Database Connection Closed")
```
